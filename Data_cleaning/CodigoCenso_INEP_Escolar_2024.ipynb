```
import pandas as pd

# ==========================================
# 1. CONFIGURAÇÕES INICIAIS
# ==========================================
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)

# ==========================================
# 2. SELEÇÃO DE COLUNAS E LEITURA DO ARQUIVO
# ==========================================
colunas_selecionadas = [
    'SG_UF', 'CO_ENTIDADE', 'IN_AGUA_POTAVEL', 'IN_AGUA_INEXISTENTE',
    'IN_ENERGIA_INEXISTENTE', 'IN_ENERGIA_REDE_PUBLICA', 'IN_ESGOTO_REDE_PUBLICA',
    'IN_ESGOTO_INEXISTENTE', 'IN_BANHEIRO', 'IN_COZINHA', 'IN_REFEITORIO',
    'IN_QUADRA_ESPORTES', 'QT_SALAS_UTILIZADAS_ACESSIVEIS', 'QT_SALAS_UTILIZADAS',
    'IN_LABORATORIO_CIENCIAS', 'IN_LABORATORIO_INFORMATICA', 'IN_SALA_DIRETORIA',
    'IN_SECRETARIA', 'IN_SALA_MULTIUSO', 'IN_SALA_LEITURA', 'IN_ALIMENTACAO',
    'IN_INTERNET', 'IN_INTERNET_ALUNOS'
]

# Caminho do arquivo (upload direto na barra lateral do Colab)
caminho_arquivo = '/content/microdados_ed_basica_2024.csv'

print("Lendo o arquivo... Isso pode demorar um pouco.")
df_censo_2024 = pd.read_csv(
    caminho_arquivo,
    sep=';',
    encoding='latin-1',
    usecols=colunas_selecionadas,
    low_memory=False
)

# ==========================================
# 3. TRATAMENTO DE NULOS (BRASIL INTEIRO)
# ==========================================
df_inep_selecionado = df_censo_2024

# Separamos apenas as colunas de dados/perguntas para checar o que está em branco
colunas_dados = [c for c in colunas_selecionadas if c not in ['SG_UF', 'CO_ENTIDADE']]

# Guarda a quantidade total inicial de escolas (linhas) ANTES da limpeza
total_escolas_antes = len(df_inep_selecionado)

# Remove APENAS as escolas (linhas) onde TODAS as colunas de dados estão vazias
df_inep_selecionado = df_inep_selecionado.dropna(subset=colunas_dados, how='all')

# Calcula a quantidade exata de linhas deletadas e o total final
linhas_excluidas_nulas = total_escolas_antes - len(df_inep_selecionado)
total_escolas_depois = len(df_inep_selecionado)

print(f"\n==========================================")
print(f"RELATÓRIO DE TRATAMENTO DE DADOS")
print(f"==========================================")
print(f"Total de escolas registradas (antes da limpeza): {total_escolas_antes}")
print(f"Total de linhas excluídas por serem 100% nulas: {linhas_excluidas_nulas}")
print(f"Total de escolas registradas (após a limpeza): {total_escolas_depois}")
print(f"==========================================\n")

# ==========================================
# 5. EXIBIÇÃO DA TABELA COMPLETA
# ==========================================
display(df_inep_selecionado.head(100))

# ==========================================
# 6. SALVAR ARQUIVO NO COMPUTADOR
# ==========================================
nome_arquivo = "censo_escolas_2024.csv"
df_inep_selecionado.to_csv(nome_arquivo, index=False, sep=';', encoding='latin-1')
```
