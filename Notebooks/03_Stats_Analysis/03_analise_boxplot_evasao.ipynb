```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df_evasao = pd.read_csv("https://raw.githubusercontent.com/MarcilioFilh0/Data_Analysis_Education/refs/heads/Data_Cleaning/src/Datas/Maiores_Taxas_Evasao_e_Reprovacao_2024.csv")

cols_to_clean = [
    'evasao_fundamental_total', 'evasao_medio_total', 'evasao_geral',
    'reprovacao_fundamental_total', 'reprovacao_medio_total', 'reprovacao_geral'
]

for col in cols_to_clean:
    df_evasao[col] = pd.to_numeric(df_evasao[col].replace('Não informado', np.nan), errors='coerce')

# ==========================================
# EVASÃO POR TIPO DE ESCOLA (%)
# ==========================================
# Calculando a Média e a Mediana para o Ensino Fundamental e Médio
tabela_resumo = df_evasao.groupby('dependencia_adm').agg(
    Média_Fundamental=('evasao_fundamental_total', 'mean'),
    Mediana_Fundamental=('evasao_fundamental_total', 'median'),
    Média_Médio=('evasao_medio_total', 'mean'),
    Mediana_Médio=('evasao_medio_total', 'median')
).round(2).reset_index()

tabela_resumo = tabela_resumo.rename(columns={'dependencia_adm': 'Tipo de Escola'})

print("--- INFORMAÇÕES DO GRÁFICO: EVASÃO POR TIPO DE ESCOLA (%) ---")
display(tabela_resumo)
print("\n" + "="*60 + "\n")


# ==========================================
# --- GERAÇÃO DOS GRÁFICOS (BOXPLOT) ---
# ==========================================
plt.figure(figsize=(12, 5))

# Gráfico 1: Distribuição no Ensino Médio
plt.subplot(1, 2, 1)
sns.boxplot(data=df_evasao, x='dependencia_adm', y='evasao_medio_total', palette='Set2', hue='dependencia_adm', legend=False)
plt.title('Distribuição da Evasão - Ensino Médio')
plt.ylabel('Taxa de Evasão (%)')
plt.xlabel('Tipo de Escola')

# Gráfico 2: Distribuição no Ensino Fundamental
plt.subplot(1, 2, 2)
sns.boxplot(data=df_evasao, x='dependencia_adm', y='evasao_fundamental_total', palette='Set3', hue='dependencia_adm', legend=False)
plt.title('Distribuição da Evasão - Ensino Fundamental')
plt.ylabel('Taxa de Evasão (%)')
plt.xlabel('Tipo de Escola')

plt.tight_layout()
plt.show()
```
